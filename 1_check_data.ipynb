```python
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
train_df = pd.read_csv('Train.csv')
```

```python
print(f'Total training images: {len(train_df)}')
print(f'Number of classes: {train_df['ClassId'].nunique()}')
```

```python
first_image_path = train_df.iloc[0]['Path']
print(f'Attempting to load image at: {first_image_path}')
```

```python
try:
    full_path = os.path.abspath(first_image_path)
    img = Image.open(full_path)
    
    plt.imshow(img)
    plt.title(f'Class ID: {train_df['ClassId'].iloc[0]}')
    plt.axis('off')
    plt.show()
    
except Exception as e:
    print(f'Error loading image {e}')
```
